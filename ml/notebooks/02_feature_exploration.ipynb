# 02 — Feature exploration
Uses the real `build_features`, so what you see here is what the model sees.

```python
import sys; sys.path.insert(0, "..")
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from src import config as C
from src.feature_engineering import build_features
raw = pd.read_csv(C.INTERIM_UPI)
X = build_features(raw); X[C.TARGET] = raw[C.TARGET]
```

```python
corr = X[C.NUMERIC_FEATURES + [C.TARGET]].corr()[C.TARGET].drop(C.TARGET).sort_values()
ax = corr.plot.barh(figsize=(6,6), title='Correlation with is_fraud')
plt.savefig(C.EDA_DIR/'feature_correlations.png', dpi=130, bbox_inches='tight')
```
