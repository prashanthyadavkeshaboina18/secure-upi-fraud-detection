# 01 — EDA
Scratch space: class balance, distributions, correlations. Saves figures to `reports/eda/`.

```python
import sys; sys.path.insert(0, "..")
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from src import config as C
df = pd.read_csv(C.INTERIM_UPI, parse_dates=['timestamp'])
df.head()
```

```python
ax = df[C.TARGET].value_counts().plot.bar(title='Class balance', rot=0)
print(df[C.TARGET].mean())
plt.savefig(C.EDA_DIR/'class_balance.png', dpi=130, bbox_inches='tight')
```

```python
fig, ax = plt.subplots(figsize=(7,4))
for lbl, g in df.groupby(C.TARGET):
    ax.hist(np.log10(g['amount']), bins=50, alpha=.6, density=True, label=f'is_fraud={lbl}')
ax.set(xlabel='log10(amount)', title='Amount distribution'); ax.legend()
fig.savefig(C.EDA_DIR/'amount_distribution.png', dpi=130, bbox_inches='tight')
```

```python
print(df[df[C.TARGET]==1]['fraud_type'].value_counts())
df.assign(hour=df.timestamp.dt.hour).groupby('hour')[C.TARGET].mean().plot(title='Fraud rate by hour')
```
